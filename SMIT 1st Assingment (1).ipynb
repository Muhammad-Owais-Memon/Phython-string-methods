# Phython String Method 

```python
name=input("Please enter your full name : ")
course=input("Please enter your favourite course :")
city=input("Please enter your city : ")
```

```python
# 1.This method converts text to uppercase .
a=name.upper()
b=course.upper()
c=city.upper()

print(a)
print(b)
print(c)
```

```python
# 2.This method converts text to lowercase .
a=name.lower()
b=course.lower()
c=city.lower()

print(a)
print(b)
print(c)
```

```python
# 3.This method converts text to titlecase .
a=name.title()
b=course.title()
c=city.title()

print(a)
print(b)
print(c)
```

```python
# 4.This method converts text to capitalizecase .
a=name.capitalize()
b=course.capitalize()
c=city.capitalize()

print(a)
print(b)
print(c)
```

```python
# 10.This method converts text to splitcase .
a=name.split()
b=course.split()
c=city.split()

print(a)
print(b)
print(c)
```
